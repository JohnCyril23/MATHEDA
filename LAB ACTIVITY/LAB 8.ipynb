```python
#imports and packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
#load dataset
df = pd.read_csv(r'C:/Users/luxah/OneDrive/MATHEDA/LAB ACTIVITY/DataSet/transformer-voltage-10-sample.csv',
                 delimiter=',',
                 index_col=["Sample"])
df
```

```python
#summary of dataframe
df.info()
```

```python
#summary of statistics
df.describe()
```

```python
# t-critical for two-tailed test
alpha = 0.05
n = df['Voltage'].count()

t_critical = stats.t.ppf(1 - alpha/2, n - 1)
t_critical
```

```python
# confidence interval
n = df['Voltage'].count()
mean = df['Voltage'].mean()
std = df['Voltage'].std(ddof=1)

ci_lower, ci_upper = stats.t.interval(
    confidence=0.95,
    df=n - 1,
    loc=mean,
    scale=std / np.sqrt(n)
)

print(f"CI: [{ci_lower:.2f} to {ci_upper:.2f}]")
```

```python
# voltage distribution
mean = df['Voltage'].mean()
std = df['Voltage'].std(ddof=1)

# x-axis
x_min = df['Voltage'].min()
x_max = df['Voltage'].max()

# normal distribution plot
x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, mean, std)

plt.plot(x, p,
         linewidth=2,
         color='#6D1616')

# histogram
plt.hist(df['Voltage'],
         bins=5,
         density=True,
         alpha=0.8,
         color='#000000')

plt.title('Voltage Distribution')
plt.ylabel('Probability Density')
plt.xlabel('Voltage (V)')

# confidence interval
plt.axvline(x=ci_lower,
            color='green',
            linestyle='--')

plt.axvline(x=ci_upper,
            color='green',
            linestyle='--')

plt.show()
```

```python
# 95% Confidence Level
mean = df.iloc[:17, 1:13].mean(axis=1)   # all 16 sizes
std = df.iloc[:17, 1:13].std(axis=1)

df['ci_lower'], df['ci_upper'] = stats.t.interval(
    confidence=0.95,
    df=n-1,
    loc=mean,
    scale=std / np.sqrt(n)
)
df
```
