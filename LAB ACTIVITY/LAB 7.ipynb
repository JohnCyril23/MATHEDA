```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

df = pd.read_csv('C:/Users/luxah/OneDrive/MATHEDA/LAB ACTIVITY/DataSet/transformer-voltage.csv',
                 index_col= 'Sample',
                 delimiter= ','
                 )

df
```

```python
#summary of statistics
df.describe()
```

```python
#z-critical for two-tailed test
alpha = 0.01
z_critical = stats.norm.ppf(1 - alpha/2)
z_critical
```

```python
#confidence interval
pop_std = 5

ci_lower, ci_upper = stats.norm.interval(
    confidence = 0.95,
    loc = df['Voltage'].mean(),
    scale = pop_std/np.sqrt(df['Voltage'].count())
)   
print(f"Confidence Interval: [{ci_lower:.2f}, {ci_upper:.2f}]")
```

```python
#voltage distribution
mean = df['Voltage'].mean()
std = df['Voltage'].std(ddof=1)

#x-axis
x_min = df['Voltage'].min()
x_max = df['Voltage'].max()

#normal distribution plot
x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, mean, std)
plt.plot(x,p,
         linewidth=2,
         color="#6D1616")

#histogram
plt.hist(df['Voltage'],
         bins=10,
         density=True,
         alpha=0.8,
         color="#000000")

plt.title('Voltage Distribution')
plt.xlabel('Voltage (V)')
plt.ylabel('Probability Density')

#confidence interval
plt.axvline(x=ci_lower,
            color='green',
            linestyle='--')
plt.axvline(x=ci_upper,
            color='green',
            linestyle='--')
plt.show()
```
